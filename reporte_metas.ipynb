```python
vendedor = "Rosa"
venta = 7420
meta = 8000
print(vendedor, "vendió", venta)
cumplimiento = venta / meta * 100
print(cumplimiento)
if cumplimiento >= 100:
    print(vendedor, "cumplió su meta")
else:
    print(vendedor, "no cumplió su meta")
```

```python
import pandas as pd

# 1. Los datos: las mismas tablas del Módulo 5
ventas = pd.DataFrame({
    "id": [1, 2, 3, 4, 5, 6, 7, 8],
    "vendedor": ["Rosa", "Carlos", "Rosa", "Lucía", "Carlos", "Lucía", "Rosa", "Carlos"],
    "zona": ["Norte", "Sur", "Norte", "Centro", "Sur", "Centro", "Norte", "Sur"],
    "producto": ["Laptop", "Mouse", "Monitor", "Laptop", "Teclado", "Monitor", "Mouse", "Laptop"],
    "cantidad": [2, 10, 3, 1, 5, 2, 8, 3],
    "precio": [2500, 40, 700, 2500, 120, 700, 40, 2500],
})

metas = pd.DataFrame({
    "vendedor": ["Rosa", "Carlos", "Lucía", "Pedro"],
    "meta": [8000, 8000, 5000, 4000],
})

# 2. Calcular el total de cada venta
ventas["total"] = ventas["cantidad"] * ventas["precio"]
print("Total vendido:", ventas["total"].sum())

# 3. Sumar por vendedor
por_vendedor = ventas.groupby("vendedor")["total"].sum().reset_index()

# 4. Unir con las metas sin perder a nadie, y poner 0 donde no hay ventas
reporte = metas.merge(por_vendedor, on="vendedor", how="left").fillna(0)

# 5. Calcular el cumplimiento y ordenar
reporte["cumplimiento"] = (reporte["total"] / reporte["meta"] * 100).round(2)
print(reporte.sort_values("cumplimiento", ascending=False))

```

```python
import numpy as np
reporte["resultado"] = np.where(reporte["cumplimiento"] >= 100, "Cumplió", "No cumplió")
print(reporte)
```
